### Advance operation with Business examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id,2021,2020,2023,2024]

sales_data = np.array([
  [1,150000,180000,220000,250000], #Paradise Biryani
  [2,120000,140000,160000,190000], #Beijing Bites
  [3,200000,230000,260000,300000], #Pizza Hub
  [4,180000,210000,240000,270000], #Burger Point
  [5,160000,185000,205000,230000]  #Chai Point
])

print("==== Zomato sales analysis ====")
print("\n Sales data shape", sales_data.shape)
print("first three row \n ", sales_data[:3])
print("All rows coloumn leaving 0 index \n ", sales_data[:,1:])
```

### total sales per year

```python
total_yearly_sales = np.sum(sales_data[:,1:],axis=0) # axis 0 = column
print(total_yearly_sales)

# minimum sales per restaurant 
min_sales = np.min(sales_data[:,1:],axis=1) 
print(min_sales)

#maximum sale per year
max_sale_yearly = np.max(sales_data[:,1:],axis=0)
print(max_sale_yearly)

#avarage sales per restaurant
avg_sales = np.average(sales_data[:,1:],axis=1)
print(avg_sales)

```

```python
#cummulative sales 
cum_sales = np.cumsum(sales_data[:,1:],axis=1)
print(cum_sales)

plt.figure(figsize=(10,6))
plt.plot(np.mean(cum_sales, axis=0))
plt.title("Average cumulative sales across all restaurant")
plt.xlabel("Years")
plt.ylabel("sales")
```
